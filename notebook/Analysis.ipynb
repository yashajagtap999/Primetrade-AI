```python
# FIND DATE COLUMNS :--

trade_date_col = "Timestamp IST"

sent_date_col = "date"
if sent_date_col not in sentiment.columns:
    sent_date_col = "timestamp"
```

```python
# CONVERT DATES :--

trades["Date"] = pd.to_datetime(
    trades[trade_date_col],
    dayfirst=True,
    errors="coerce"
).dt.date

sentiment["Date"] = pd.to_datetime(
    sentiment[sent_date_col],
    errors="coerce"
).dt.date
```

```python
# FIND CLASSIFICATION COLUMN :--

classification_col = None

for col in sentiment.columns:
    if "class" in col.lower():
        classification_col = col
        break

print("\nClassification Column:", classification_col)
```

```python
# MERGE :--

merged = trades.merge(
    sentiment[["Date", classification_col]],
    on="Date",
    how="left"
)

merged.rename(
    columns={classification_col: "Classification"},
    inplace=True
)

print("\nMerged Shape:", merged.shape)
```

```python
# CHECK MERGE QUALITY :--
print("\nClassification Counts")
print(
    merged["Classification"].value_counts(
        dropna=False
    )
)
```

```python
# FIND PNL COLUMN :--
pnl_col = None

for col in merged.columns:
    if "pnl" in col.lower():
        pnl_col = col
        break

print("\nPnL Column:", pnl_col)
```

```python
# WIN RATE :--

merged["Winning_Trade"] = merged[pnl_col] > 0

win_rate = (
    merged.groupby("Classification")
    ["Winning_Trade"]
    .mean()
    * 100
)

print("\nWin Rate")
print(win_rate)
```

```python
# PERFORMANCE :--

performance = merged.groupby(
    "Classification"
).agg(
    Total_Trades=(pnl_col, "count"),
    Average_PnL=(pnl_col, "mean"),
    Total_PnL=(pnl_col, "sum"),
    Median_PnL=(pnl_col, "median")
)

print("\nPerformance")
print(performance)
```

```python
# INSIGHTS :--
if len(performance) > 0:

    print("\nKEY FINDINGS")

    print(
        "Highest Average PnL:",
        performance["Average_PnL"].idxmax()
    )

    print(
        "Highest Total PnL:",
        performance["Total_PnL"].idxmax()
    )

    print(
        "Highest Win Rate:",
        win_rate.idxmax()
    )

else:
    print(
        "\nNo sentiment data matched. Check date merge."
    )
```

```python
# POSITION SIZE ANALYSIS :--


size_cols = [c for c in merged.columns if 'size' in c.lower()]

print("Size Columns:", size_cols)

if len(size_cols) > 0:

    size_col = size_cols[0]

    position_analysis = merged.groupby(
        'Classification'
    )[size_col].mean()

    print("\nAverage Position Size")
    print(position_analysis)

    position_analysis.plot(
        kind='bar',
        figsize=(8,5)
    )

    plt.title(
        'Average Position Size by Sentiment'
    )

    plt.ylabel('Position Size')

    plt.show()
```

```python
# SIDE ANALYSIS :--
side_cols = [
    c for c in merged.columns
    if 'side' in c.lower()
]

print("Side Columns:", side_cols)

if len(side_cols) > 0:

    side_col = side_cols[0]

    side_analysis = pd.crosstab(
        merged['Classification'],
        merged[side_col]
    )

    print("\nBuy vs Sell Analysis")
    print(side_analysis)

    side_analysis.plot(
        kind='bar',
        figsize=(10,5)
    )

    plt.title(
        'Buy/Sell Behaviour by Sentiment'
    )

    plt.show()
```

```python
# MOST TRADED COINS :--

coin_cols = [
    c for c in merged.columns
    if c.lower() in ['coin','symbol']
]

if len(coin_cols) > 0:

    coin_col = coin_cols[0]

    print("\nTop Coins")

    print(
        merged[coin_col]
        .value_counts()
        .head(10)
    )
```

```python
# TOP TRADERS BY SENTIMENT :--

account_cols = [
    c for c in merged.columns
    if 'account' in c.lower()
]

if len(account_cols) > 0:

    account_col = account_cols[0]

    trader_sentiment = merged.groupby(
        ['Classification', account_col]
    )['Closed PnL'].sum()

    print(
        trader_sentiment
        .sort_values(ascending=False)
        .head(20)
    )
```
